### Limpandon a base

```python
def limpar_nome_coluna_v2(col):
    col_str = str(col).strip()

    # 1. Se já for uma tupla real do Python, pega o segundo elemento
    if isinstance(col, tuple) and len(col) >= 2:
        nome = col[1]

    # 2. Se for uma string que parece tupla malformada: "('cod', 'texto com apóstrofo's')"
    elif col_str.startswith("('") or col_str.startswith('("'):
        if "', '" in col_str:
            # Separa na primeira ocorrência de "', '" e pega a segunda parte, removendo o final "')"
            nome = col_str.split("', '", 1)[1].rstrip("')")
        elif '", ' in col_str:
            nome = col_str.split('", ', 1)[1].rstrip('")')
        else:
            nome = col_str # Fallback de segurança

    # 3. Se for um nome normal, mantém
    else:
        nome = col_str

    # --- LIMPEZA PADRÃO DO TEXTO ---
    nome = nome.strip().lower()

    # Substitui caracteres problemáticos por underline ou remove
    nome = nome.replace(" ", "_").replace("/", "_").replace("-", "_")
    nome = nome.replace("?", "").replace("(", "").replace(")", "")
    nome = nome.replace("'", "").replace('"', "") # Remove os apóstrofos que sobrarem (ex: etls em vez de etl's)

    # Remove acentos de forma segura e profissional
    nome = ''.join(c for c in unicodedata.normalize('NFD', nome) if unicodedata.category(c) != 'Mn')

    return nome

# Aplica a nova função em todas as colunas
df_2023.columns = [limpar_nome_coluna_v2(c) for c in df_2023.columns]

# Verifica se ficou limpo (procure por 'etl' ou 'api' agora)
print("Colunas que antes estavam quebradas:")
print([c for c in df_2023.columns if 'etl' in c or 'api' in c or 'hmm' in c])
```

```python
def finalizar_silver(df, ano):
    print(f"\n===== {ano} =====")

    # --- 1) Checar linhas duplicadas (pelo id) ---
    dup = df['id'].duplicated().sum()
    print(f"IDs duplicados: {dup} | IDs nulos: {df['id'].isnull().sum()}")
    if dup > 0:
        df = df.drop_duplicates(subset=['id'])   # mesma pessoa 2x → mantém 1
        print("✔ Duplicatas removidas pelo id")

    # --- 2) Binárias (0/1): nulo vira 0 ("não marcou / não usa") ---
    binarias = [c for c in df.columns if set(df[c].dropna().unique()) <= {0, 1}]
    df[binarias] = df[binarias].fillna(0).astype(int)

    # --- 3) Texto/categoria: nulo vira 'Nao_informado' (ninguém some dos gráficos!) ---
    texto = [c for c in df.select_dtypes(include='object').columns if c != 'id']
    df[texto] = df[texto].fillna('Nao_informado')

    # --- 4) Tag do ano (para análise temporal!) ---
    df['ano'] = ano

    print(f"Shape final: {df.shape} | NaNs restantes: {df.isnull().sum().sum()}")
    return df

df_silver_2023 = finalizar_silver(df_silver_2023, 2023)
df_silver_2024 = finalizar_silver(df_silver_2024, 2024)
df_silver_2025 = finalizar_silver(df_silver_2025, 2025)
```
